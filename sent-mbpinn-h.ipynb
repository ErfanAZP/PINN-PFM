```python
"""
Implements energy-based PINN (deep energy method) for hydrogen-assisted phase-field fracture.

Solves the coupled displacement and phase fields by directly minimizing the
regularized potential energy, with the fracture toughness locally degraded by
hydrogen. The hydrogen concentration is obtained algebraically from the hydrostatic
stress under the Oriani equilibrium (fast-diffusion) assumption, so no separate
transport equation is solved; coverage and toughness degradation follow a
Langmuir--McLean isotherm.

Key assumptions:
  - Quasi-static loading; hydrogen stays in equilibrium with the stress field.
  - Pre-charged specimen at a uniform nominal concentration (C0_wt_ppm within the code).
  - Toughness degradation saturates at high coverage (single-trap isotherm).

Boundary conditions are imposed exactly through the network ansatz. Training uses Adam
followed by full-grid L-BFGS; the energy integral may optionally be estimated from
random minibatches of the integration points (set batch_frac; batch_frac = 1
recovers full-batch training).
"""

import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import tri as mtri
import time
import copy
import pickle, os

np.random.seed(1234)
torch.manual_seed(1234)
torch.set_default_dtype(torch.float64)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
#device = torch.device("cpu")  # Force CPU for debugging
print("Using device:", device)
```

```python
E   = 210e3
nu  = 0.3
Gc  = 2.7

# Hydrogen parameters 
chi      = 0.89                  # HEDE damage coefficient (iron-based)
R_gas    = 8.314e3               # universal gas constant [N·mm / (mol·K)]
T        = 300.0                 # temperature [K]
delta_g  = 30.0e6                # Gibbs free energy of trapping [N·mm/mol]
V_H   = 2000.0                   # mm³/mol
alpha = V_H / (R_gas * T)

# Hydrogen concentration (mole fraction)
C0_wt_ppm = 0.5                 # Parameter sweeps for 0.0, 0.1, 0.5, 1.0 
C0_molfrac = C0_wt_ppm * 5.55e-5

L = 1.0
W = 1.0
xmin, xmax = 0.0, L
ymin, ymax = 0.0, W
ls = 0.0075
plane_strain = True
use_split = True

lam = E*nu / ((1+nu)*(1-2*nu)) if plane_strain else E*nu / (1-nu**2)
mu  = E / (2*(1+nu))

# Incremental loading
nSteps = 20
deltaV = 2.e-4
v_deltas = (np.arange(0, nSteps, dtype=np.float64) * deltaV)

# Training parameters
n_epochs_step0 = 20000
n_epochs_other = 4000
lbfgs_step0_iter = 6000
lbfgs_other_iter = 1000
print_every = 1000
plot_every = 1
```

```python
def gen_grid(L, W, ls):
    inner_offset = 3.0 * ls   # half-width of the fine strip
    outer_offset = 6.0 * ls   # half-width of the transition strip

    sec_bounds = np.array([
        [0.0,                  0.5*W - outer_offset],
        [0.5*W - outer_offset, 0.5*W - inner_offset],  
        [0.5*W - inner_offset, 0.5*W + inner_offset], 
        [0.5*W + inner_offset, 0.5*W + outer_offset],
        [0.5*W + outer_offset, W]])                   

    nPred = np.array([
        [50,  20],   
        [70, 20],     
        [140, 25],  
        [70, 20],   
        [50,  20]])  

    XY_list, w_list = [], []

    for i in range(5):
        nx, ny = nPred[i]
        y1, y2 = sec_bounds[i]

        xs = np.linspace(0.0, L, nx)
        ys = np.linspace(y1, y2, ny, endpoint=(i == 4))

        X, Y = np.meshgrid(xs, ys)
        XY = np.hstack([X.reshape(-1,1), Y.reshape(-1,1)])
        XY_list.append(XY)

        dx = L / (nx - 1)
        dy = (y2 - y1) / (ny - 1) if ny > 1 else (y2 - y1)
        w = np.full((XY.shape[0],), dx*dy, dtype=np.float64)
        w_list.append(w)

    Grid = np.vstack(XY_list).astype(np.float64)
    w_f  = np.hstack(w_list).astype(np.float64)

    return Grid, w_f
```

```python
B = 1000.0

X_f, w_f = gen_grid(L, W, ls)
Nq = X_f.shape[0]
xg_t = torch.from_numpy(X_f).to(device=device, dtype=torch.get_default_dtype())
wg_t = torch.from_numpy(w_f).to(device=device, dtype=torch.get_default_dtype())

x = X_f[:,0].astype(np.float64)
y = X_f[:,1].astype(np.float64)

x_tip, y_tip = 0.5, 0.5
crackTip = x_tip
dist = np.zeros((Nq,1), dtype=np.float64)
mask_ahead = x > crackTip
mask_behind = ~mask_ahead
dist[mask_behind, 0] = np.abs(y[mask_behind] - y_tip)
dist[mask_ahead,  0] = np.sqrt((x[mask_ahead]-x_tip)**2 + (y[mask_ahead]-y_tip)**2)
inside = dist[:,0] < 0.5*ls

H_init_np = np.zeros((Nq,1), dtype=np.float64)
H_init_np[inside, 0] = B*Gc*0.5*(1.0 - 2.0*dist[inside,0]/ls)/ls
H_init = torch.from_numpy(H_init_np).to(device=device, dtype=torch.get_default_dtype())

#  Initialize H from H_init
H = torch.maximum(torch.zeros_like(H_init), H_init.clone())

phi0 = np.zeros((Nq,1), dtype=np.float64)
phi0_t = torch.from_numpy(phi0).to(device=device, dtype=torch.get_default_dtype())
```

```python
def strain_calc(u, x):

    ux, uy = u[:, 0:1], u[:, 1:2]

    grad_ux = torch.autograd.grad(ux, x, torch.ones_like(ux), create_graph=True)[0]
    grad_uy = torch.autograd.grad(uy, x, torch.ones_like(uy), create_graph=True)[0]

    ux_x = grad_ux[:, 0:1]
    ux_y = grad_ux[:, 1:2]
    uy_x = grad_uy[:, 0:1]
    uy_y = grad_uy[:, 1:2]

    eps_xx = ux_x
    eps_yy = uy_y
    eps_xy = 0.5*(ux_y + uy_x)

    return eps_xx, eps_yy, eps_xy

def grad_phi(phi, x):
    gphi = torch.autograd.grad(phi, x, torch.ones_like(phi), create_graph=True)[0]
    return gphi[:,0:1], gphi[:,1:2]

def princ_strains_2d(eps_xx, eps_yy, eps_xy, eps=1e-12):
    tr = eps_xx + eps_yy
    diff = eps_xx - eps_yy
    disc = torch.sqrt(diff**2 + 4.0*eps_xy**2 + 1e-16) 
    e1 = 0.5*(tr + disc)
    e2 = 0.5*(tr - disc)
    return e1, e2

def el_energy(eps_xx, eps_yy, eps_xy, lam, mu):
    tr = eps_xx + eps_yy
    return 0.5*lam*tr**2 + mu*(eps_xx**2 + eps_yy**2 + 2*eps_xy**2)

def el_energy_spectral_split(eps_xx, eps_yy, eps_xy, lam, mu):
    
    e1, e2 = princ_strains_2d(eps_xx, eps_yy, eps_xy)
    tr = e1 + e2

    trp = torch.relu(tr)
    trm = torch.relu(-tr)

    e1p, e2p = torch.relu(e1),  torch.relu(e2)
    e1m, e2m = torch.relu(-e1), torch.relu(-e2)

    Psi_pos = 0.5*lam*trp**2 + mu*(e1p**2 + e2p**2)
    Psi_neg = 0.5*lam*trm**2 + mu*(e1m**2 + e2m**2)

    return Psi_pos, Psi_neg
```

```python
def build_mlp(layers, activation=nn.Tanh):
    modules = []
    for i in range(len(layers)-1):
        modules.append(nn.Linear(layers[i], layers[i+1]))
        if i < len(layers)-2:
            modules.append(activation())
    net = nn.Sequential(*modules)
    for m in net.modules():
        if isinstance(m, nn.Linear):
            nn.init.xavier_normal_(m.weight)
            nn.init.zeros_(m.bias)
    return net

class PINN2D_PF(nn.Module):
    def __init__(self, layers):
        super().__init__()
        self.net = build_mlp(layers)

    def forward_raw(self, x):
        lb = torch.tensor([xmin, ymin], device=x.device, dtype=x.dtype)
        ub = torch.tensor([xmax, ymax], device=x.device, dtype=x.dtype)
        x_n = 2.0*(x - lb)/(ub - lb) - 1.0
        return self.net(x_n)

    def apply_BC(self, x, v_delta, out_raw):

        """
        BC definition based on Goswami et al. (2020)
        """
        ux_hat = out_raw[:,0:1]
        uy_hat = out_raw[:,1:2]
        phi_hat = out_raw[:,2:3]

        xhat = (x[:,0:1] - xmin)/(xmax - xmin)
        yhat = (x[:,1:2] - ymin)/(ymax - ymin)

        # Enforce ux=0 on both left and right boundaries
        ux = xhat*(1.0 - xhat)*ux_hat

        # Enforce uy=0 at y=0, uy=v_delta at y=1
        uy = yhat*(yhat - 1.0)*uy_hat + yhat*v_delta

        phi = phi_hat
 #       phi = 0.5*(torch.tanh(phi_hat) + 1.0)
 #       phi= torch.sigmoid(phi_hat)
 #       phi = torch.clamp(phi_hat, 0, 1)

        return ux, uy, phi

    def apply_BC2(self, x, v_delta, out_raw):

       ux_hat  = out_raw[:, 0:1]
       uy_hat  = out_raw[:, 1:2]
       phi_hat = out_raw[:, 2:3]

        # Centered coordinates: (xc, yc)
       half_Lx = 0.5 * (xmax - xmin)   
       half_Ly = 0.5 * (ymax - ymin)  
       xc = x[:, 0:1] - (xmin + half_Lx)  
       yc = x[:, 1:2] - (ymin + half_Ly)

        #  Hard BC: ux, u = (x_c^2 - 0.25) * û
        # Enforces ux = 0 at left (x=0) and right (x=1)
       ux = (xc**2 - half_Lx**2) * ux_hat

        #  Hard BC: uy, v = (y_c^2 - 0.25)*v̂ + (y_c + 0.5)*Δv
       uy = (yc**2 - half_Ly**2) * uy_hat + (yc + half_Ly) * v_delta

       phi = phi_hat
    #   phi = 0.5*(torch.tanh(phi_hat) + 1.0)
    #   phi= torch.sigmoid(phi_hat)

       return ux, uy, phi

    def forward(self, x, v_delta):
           out_raw = self.forward_raw(x)
           return self.apply_BC2(x, v_delta, out_raw)
```

```python
def hydrostatic_stress(model_uphi, x, v_delta, lam, mu, nu):
    """
    Compute σ_H at points x using the (u,φ) network.
    Uses plane-strain assumption!
    """
    x = x.requires_grad_(True)
    ux, uy, phi = model_uphi(x, v_delta)
    u = torch.cat([ux, uy], dim=1)
    eps_xx, eps_yy, eps_xy = strain_calc(u, x)
    
    kappa = 1e-7
    g = (1.0 - phi)**2 + kappa
    
    tr_eps = eps_xx + eps_yy
    sigma_xx = g*(lam * tr_eps + 2.0 * mu * eps_xx)
    sigma_yy = g*(lam * tr_eps + 2.0 * mu * eps_yy)
    
    # Plane strain: σ_zz = ν(σ_xx + σ_yy)
    # σ_H = (σ_xx + σ_yy + σ_zz)/3 = (1+ν)/3 · (σ_xx + σ_yy)
    sigma_H = (1.0 + nu) / 3.0 * (sigma_xx + sigma_yy)
    return sigma_H

def hydrogen_coverage(C, delta_g, R_gas, T):
    """
    Langmuir-McLean isotherm: convert bulk hydrogen concentration (mole fraction)
    to surface coverage θ ∈ [0, 1] 
    """
    denom = C + np.exp(-delta_g / (R_gas * T))
    return C / denom

def hede_Gc(C, Gc0, chi, delta_g, R_gas, T):
    """
    HEDE law: Gc(C)/Gc(0) = 1 - χ·θ(C)
    """
    if isinstance(C, torch.Tensor):
        denom = C + np.exp(-delta_g / (R_gas * T))
        theta = C / denom
    else:
        theta = hydrogen_coverage(C, delta_g, R_gas, T)
    return Gc0 * (1.0 - chi * theta)

C_field = torch.full((Nq, 1), C0_molfrac,
                     dtype=torch.get_default_dtype(),
                     device=device)
Gc_field = torch.full((Nq, 1), Gc,
                       dtype=torch.get_default_dtype(), device=device)
```

```python
def uniform_sample(x_all, w_all, n_batch):
    """
    Uniform random subsampling of the graded grid. Returns (x_batch, w_batch, idx)
    where w_batch is corrected so that sum(w_batch * psi) is an unbiased estimator
    of the full integral sum(w_all * psi).
    """
    N = x_all.shape[0]
    idx = torch.randint(0, N, (n_batch,), device=x_all.device)
    x_batch = x_all[idx]
    w_batch = w_all[idx] * (N / n_batch)
    return x_batch, w_batch, idx
```

```python
def energy_and_history(model, x, wg_t, H_prev, v_delta, Gc_local=None, H_init_local=H_init):
    x = x.requires_grad_(True)
    ux, uy, phi = model(x, v_delta)
    u = torch.cat([ux, uy], dim=1)
    
    eps_xx, eps_yy, eps_xy = strain_calc(u, x)
    
    # Elastic energy density with or without spectral split based on Miehe et al. (2010)
    if use_split:
        Psi_pos, Psi_neg = el_energy_spectral_split(eps_xx, eps_yy, eps_xy, lam, mu)
        Psi_drv = Psi_pos
    else:
        Psi_drv = el_energy(eps_xx, eps_yy, eps_xy, lam, mu)
        Psi_neg = torch.zeros_like(Psi_drv)
    
    # Irreversible history field
    H_new = torch.maximum(torch.maximum(Psi_drv, H_prev), H_init_local)
    
    # Phase-field gradient
    dphix, dphiy = grad_phi(phi, x)
    nabla_phi_sq = dphix**2 + dphiy**2
    
    # Degradation function
    kappa = 1e-7
    g = (1.0 - phi)**2 + kappa
    
    # Total energy density: elastic + fracture
    if use_split:
        psi_elastic = g * Psi_pos + Psi_neg
    else:
        psi_elastic = g * Psi_drv
    
    psi_fracture = 0.5 * Gc_local * (phi**2 / ls + ls * nabla_phi_sq) + g * H_new
    
    E_tot = torch.sum((psi_elastic + psi_fracture).squeeze() * wg_t)
    
    return E_tot, H_new.detach()
```

```python
phi_snapshots = []
C_snapshots   = []
Gc_snapshots  = []
sigma_H_snapshots = []
load_vals     = []
loss_history  = {}
phi_prev      = phi0_t.clone()

peak_force = -np.inf
peak_step  = 0

step_times   = {}
total_start  = time.time()

# Networks — created once, persisted across load steps
model   = PINN2D_PF(layers=[2, 50, 50, 50, 3]).to(device).double()

# Whether to actually solve diffusion (skip when C0=0 to save time)
do_diffusion = (C0_wt_ppm > 0.0)

batch_frac = 4
n_batch = X_f.shape[0] // batch_frac

for istep, v_val in enumerate(v_deltas):

    step_start = time.time()
    v_delta = torch.tensor(v_val, dtype=torch.get_default_dtype(), device=device)
    n_epochs = n_epochs_step0 if istep == 0 else n_epochs_other
    max_iter = lbfgs_step0_iter if istep == 0 else lbfgs_other_iter
    print(f"\n=== Step {istep}/{len(v_deltas)-1} | v_delta={v_val:.3e} ===")

    # ────────────────────────────────────────────────────────────────
    #  MECHANICS + PHASE-FIELD with current Gc_field
    # ────────────────────────────────────────────────────────────────
    for param in model.parameters():
        param.requires_grad = True
    model.train()

    adam_lr   = 1e-3 if istep == 0 else 1e-4
    optimizer = torch.optim.Adam(
        filter(lambda p: p.requires_grad, model.parameters()), lr=adam_lr)
    lbfgs = torch.optim.LBFGS(
        filter(lambda p: p.requires_grad, model.parameters()),
        lr=1.0, max_iter=max_iter,
        tolerance_grad=1e-9, tolerance_change=1e-12,
        line_search_fn='strong_wolfe')

    loss_history[istep] = {'adam_iters': [], 'adam_losses': [],
                           'lbfgs_iters': [], 'lbfgs_losses': []}

    x = xg_t.detach().clone().requires_grad_(True)

    ## ADAM OPTIMIZATION

    adam_start = time.time()

    for epoch in range(1, n_epochs + 1):
    # minibatch sampling
        x_batch, w_batch, idx = uniform_sample(xg_t, wg_t, n_batch)
        x_batch = x_batch.detach().clone().requires_grad_(True)

        optimizer.zero_grad()
        E_tot, H_new = energy_and_history(model, x_batch, w_batch, H[idx], v_delta,
                                          Gc_local=Gc_field[idx], H_init_local=H_init[idx])   
        E_tot.backward()
        optimizer.step()
        loss_history[istep]['adam_iters'].append(epoch)
        loss_history[istep]['adam_losses'].append(E_tot.item())
        if epoch % print_every == 0:
            with torch.no_grad():
                _, _, phi_dbg = model(xg_t, v_delta)
            print(f"  Adam {epoch:5d} | E={E_tot.item():.4e}"
                  f" | avg(phi)={phi_dbg.mean().item():.3e}"
                  f" | Gc min={Gc_field.min().item():.3f}"
                  f" | t={time.time()-adam_start:.1f}s")

    lbfgs_iter_count = [0]
    lbfgs_start      = time.time()

    def closure():
        lbfgs.zero_grad()
        E, _ = energy_and_history(model, x, wg_t, H, v_delta, Gc_local=Gc_field)
        E.backward()
        if lbfgs_iter_count[0] % print_every == 0:
            with torch.no_grad():
                _, _, phi_lbfgs = model(xg_t, v_delta)
            print(f"  LBFGS {lbfgs_iter_count[0]:5d} | E={E.item():.4e}"
                  f" | avg(phi)={phi_lbfgs.mean().item():.3e}"
                  f" | t={time.time()-lbfgs_start:.1f}s")
        lbfgs_iter_count[0] += 1
        loss_history[istep]['lbfgs_iters'].append(lbfgs_iter_count[0])
        loss_history[istep]['lbfgs_losses'].append(E.item())
        return E

    lbfgs.step(closure)

    # Update history field
    x_final = xg_t.detach().clone().requires_grad_(True)
    _, H_new = energy_and_history(model, x_final, wg_t, H, v_delta,
                                  Gc_local=Gc_field)
    H = H_new.detach()

    with torch.no_grad():
        _, _, phi_raw_n = model(grid_t, v_delta)
        phi_n    = torch.maximum(phi_raw_n, phi_prev)
        phi_prev = phi_n.clone()

    # ────────────────────────────────────────────────────────────────
    # (b) HYDROGEN DIFFUSION 
    # ────────────────────────────────────────────────────────────────
    sigma_threshold = 200.0  # N/mm² — minimum σ_H_max 
    C_peak_history = {} 

    with torch.enable_grad():
        xc = xg_t.detach().clone().requires_grad_(True)
        sH_peek = hydrostatic_stress(model, xc, v_delta, lam, mu, nu)
    sH_max = sH_peek.max().item()

    if do_diffusion and istep > 0 and sH_max > sigma_threshold:
        # Equilibrium (Oriani) hydrogen distribution — fast-diffusion limit.
        with torch.no_grad():
            sH       = sH_peek.detach()
            C_tilde  = torch.exp(alpha * sH)            
            C_abs    = C_tilde * C0_molfrac
            Gc_field = hede_Gc(C_abs, Gc, chi, delta_g, R_gas, T)
            C_abs_ppm = C_tilde * C0_wt_ppm  
            print(f"  C̃    range: [{C_tilde.min().item():.3f}, {C_tilde.max().item():.3f}]")
            print(f"  Gc   range: [{Gc_field.min().item():.4f}, {Gc_field.max().item():.4f}] N/mm")
            print(f"  Gc reduction at tip: {100*(1 - Gc_field.min().item()/Gc):.1f}%")
        C_snapshots.append(C_tilde.cpu().numpy().ravel())
        Gc_snapshots.append(Gc_field.cpu().numpy().ravel())

        C_np = C_abs.cpu().numpy().ravel()
        crack = (np.abs(X_f[:,1] - 0.5) < 0.5*ls) & (X_f[:,0] < 0.5)
        C_peak = C_np[~crack].max()                   
        C_peak_history.setdefault(C0_wt_ppm, []).append((v_val, C_peak))
    else:
        C_snapshots.append(np.ones(Nq))      
        Gc_snapshots.append(np.full(Nq, Gc))
        C_peak_history.setdefault(C0_wt_ppm, []).append((v_val, C0_wt_ppm))

    phi_snapshots.append(phi_n.detach().cpu().numpy().ravel())
    load_vals.append(v_val)


    if do_diffusion and istep > 0 and istep % plot_every == 0:
        with torch.enable_grad():
            x_sh = grid_t.detach().clone().requires_grad_(True)
            sigma_H_field = hydrostatic_stress(model, x_sh, v_delta, lam, mu, nu)
        sigma_H_np = sigma_H_field.detach().cpu().numpy().ravel()

        crack   = (np.abs(X_f[:,1] - 0.5) < 0.5*ls) & (X_f[:,0] < 0.5)
        C_disp  = C_snapshots[-1].copy();   C_disp[crack]  = 1.0
        Gc_disp = Gc_snapshots[-1].copy()
        sH_disp = sigma_H_np.copy();        sH_disp[crack] = 0.0
        phi_disp = phi_snapshots[-1].copy()

      #  plot_vars(C_disp, sH_disp, phi_disp, istep=istep, v_val=v_val)
        plot_vars2(phi_disp, C_disp, Gc_disp, sH_disp, istep=istep, v_val=v_val)
        
    # Reaction force
    force = compute_reaction_force(model, xy_top_t, v_delta, lam, mu, L, N_b)
    force_value = -force.item()
    fdGraph[istep, 0] = v_val
    fdGraph[istep, 1] = force_value

    if force_value > peak_force:
        peak_force = force_value
        peak_step  = istep

    if peak_force > 50 and force_value < 0.1 * peak_force:
        print(f"\nFull fracture reached at step {istep} (F={force_value:.1f}, peak was {peak_force:.1f}). Stopping.")
    # Trim arrays to actual steps run
        fdGraph = fdGraph[:istep+1]
        break

    step_times[istep] = time.time() - step_start
    print(f"  Step {istep} time: {step_times[istep]:.1f}s")

fd = fdGraph[:len(load_vals)]

plot_force_disp(fdGraph)
plot_loss(loss_history, step=0)
print(f"\nTotal training time: {time.time() - total_start:.1f}s")


os.makedirs('/kaggle/working/hydro', exist_ok=True)
with open(f'/kaggle/working/hydro/peakC_{C0_wt_ppm}.pkl', 'wb') as f:
    pickle.dump({'C0': C0_wt_ppm,
                 'u':      [d[0] for d in C_peak_history[C0_wt_ppm]],
                 'C_peak': [d[1] for d in C_peak_history[C0_wt_ppm]],
                 'fd':     fdGraph.copy()}, f)   
print(f"saved C0={C0_wt_ppm}")
```
